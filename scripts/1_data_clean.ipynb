```python
# Script: 1_data_clean.ipynb
# Project: kenya_coastal_climate_plans
# Purpose: lists number of women and youth in the six coastal counties from 2019 census into one clean, analysis-ready file
# Youth : Ages 18-34
# Reads:   raw_data/kenya_census_2019_age_sex_county.csv
# Writes:  clean_data/census_2019_coastal_counties_clean.csv
# setting the working directory
setwd('/content/drive/MyDrive/kenya_coastal_climate_plans')
# for reading and saving CSV files
library(readr)
# for cleaning
library(dplyr)
getwd()
```

```python
census_raw <- read_csv('raw_data/kenya_census_2019_age_sex_county.csv')
glimpse(census_raw)
#examining my outputs
unique(census_raw$County)
unique(census_raw$SubCounty)
unique(census_raw$Age)
```

```python
# Keep the six coastal counties, using whole-county rows only (SubCounty = ALL)
coastal <- census_raw %>%
  filter(County %in% c('LAMU', 'MOMBASA', 'KILIFI', 'TANA RIVER', 'TAITA/ TAVETA', 'KWALE'), SubCounty == 'ALL')
unique(coastal$County)
```

```python
# A table of population for each coastal county with the percentage of women
# Use only the Age = Total row, because it gives the total population
sex_by_county <- coastal %>%
  filter(Age == 'Total') %>%
  select(County, Male, Female, Total) %>%
  mutate(women_percentage = Female/Total * 100)
sex_by_county
```

```python
# A table with one row per county showing the number of youth
youth_per_county <- coastal %>%
  filter(Age %in%  as.character(18:34)) %>%
  group_by(County) %>%
  summarise(youth = sum(Total))
youth_per_county
```

```python
# Joining our tables
coastal_summary <- sex_by_county %>%
  left_join(youth_per_county, by = "County") %>%
  mutate(youth_percentage = youth / Total * 100)
coastal_summary
```

```python
# Saving coastal_summary to clean data
write_csv(coastal_summary, 'clean_data/census_2019_coastal_counties_clean.csv')
list.files('clean_data')
```
